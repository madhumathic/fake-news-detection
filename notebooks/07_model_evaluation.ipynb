```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()   # project root
sys.path.insert(0, str(ROOT))
import warnings; warnings.filterwarnings("ignore")
import pandas as pd
pd.set_option("display.max_colwidth", 80)
```

```python
import joblib
from src.config import MODEL_PATH, VECTORIZER_PATH, TRAIN_PATH, TEST_PATH
from src.models.evaluate import evaluate_on_test
model, vec = joblib.load(MODEL_PATH), joblib.load(VECTORIZER_PATH)
train, test = pd.read_csv(TRAIN_PATH), pd.read_csv(TEST_PATH)
report, y_pred, proba = evaluate_on_test(model, vec, train, test)
for k in ['accuracy', 'accuracy_95ci', 'precision', 'recall', 'f1_score', 'roc_auc', 'train_accuracy', 'majority_baseline_accuracy', 'p_value_vs_baseline']:
    print(f'{k:28}', report[k])
```
