# Phase 6 - Feature extraction (TF-IDF)

Convert text to numbers with TF-IDF. The vectorizer is fitted on the training set only.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()   # project root
sys.path.insert(0, str(ROOT))
import warnings; warnings.filterwarnings("ignore")
import pandas as pd
pd.set_option("display.max_colwidth", 80)
```

```python
from src.config import TRAIN_PATH, TEST_PATH
train = pd.read_csv(TRAIN_PATH); test = pd.read_csv(TEST_PATH)
from src.features.tfidf_features import fit_vectorizer
vec, X_train = fit_vectorizer(train['news'])
X_test = vec.transform(test['news'])   # transform only: no refitting
print(X_train.shape, X_test.shape)
print('non-zero share:', round(X_train.nnz / (X_train.shape[0] * X_train.shape[1]), 4))
```

```python
import numpy as np
terms = np.array(vec.get_feature_names_out())
idf = vec.idf_
print('most distinctive (high idf):', terms[np.argsort(idf)[::-1][:10]].tolist())
print('most common (low idf):', terms[np.argsort(idf)[:10]].tolist())
```
